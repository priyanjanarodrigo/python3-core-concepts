## <span style="color: green;">2. How many ways are there to execute a Python program</span>
<hr/>

### 2.1 Python provides a REPL tool (REPL = Read, Evaluate, Print, Loop): Python IDLE

- Search for and open **Python IDLE**. The Python shell will open.

- Alternatively, open the terminal and type `python3`. The Python shell will open in the terminal.

- Python IDLE is similar to JShell in Java.

- You can write and execute Python code directly there (for example: `print('Hello World')`).

- The code you type is read, evaluated, and printed, and then the prompt appears again for the next command.

- The output will be `Hello World`, and IDLE will be ready to accept the next line.

- Example output (macOS Terminal):

    ```text
    Python 3.13.0 (v3.13.0:60403a5409f, Oct  7 2024, 00:37:40) [Clang 15.0.0 (clang-1500.3.9.4)] on darwin
    Type "help", "copyright", "credits" or "license" for more information.
    >>> print('Hello World')
    Hello World
    >>>
    ````

- exit() function will stop python shell

###  2.2 Writing programs in separate files (ex : myProgram.py)

- if our program is larger, it is not recommended to use Python shell, and we should save the code in a separate file with <b>extension .py</b> .Then we can execute that program.

- Sample Program 1 : `demo.py`

`````python
# Objective : Read some int value from keyboard and then print a message that number of times

# Takes the input from the user and prints it using a for loop
num = int(input("Enter some number : "))

# for loop to print the value num times
for i in range(num):
    print(f'Hello index {i} ! getting errors is very common.Open Google to resolve')
`````

- Sample program 2: `calculations.py`

`````python
# calculations.py
# Takes two inputs from user and converts them to integers
x = int(input("Enter First Number : "))
y = int(input("Enter Second Number : "))
print(f'First Number is : {x} and the Second Number is : {y}')

# Prints out addition, multiplication, division outputs
print(f'x + y = {x + y}')
print(f'x * y = {x * y}')
print(f'x / y = {x / y}')
`````

### 2.3 Executing the program(s) using Python shell

- On Python IDLE, File → `New File` → Write the code content → Give a name and save with .py extension (ex : demo.py)
- On Python IDLE, File → `Open` → Select the file → `Run`

### 2.4 Executing the program(s) using command prompt/terminal

- Navigate to the file location and execute the file using the command(s) depending on your system configuration

```text
    demo.py / python demo.py
````

```text
    calculations.py / python calculations.py
````

## <span style="color: green;">3. Python IDEs and Text Editors</span>
<hr/>

- IDEs : PyCharm, Eclipse, Spyder, Visual Studio Code etc.

- Text Editors : Notepad ++, EditPlus, Sublime Text, Atom etc.

- It is not recommended to use the classic/normal Notepad since indentations are that much important in python

- From learning point of view, it is recommended to start with text editors and then move to IDEs. Once you are comfortable with the syntax and basics of Python, then you can start using IDEs. (Text Editors for learning Core Python, Advanced Python and IDEs for learning : Frameworks (Django, Flask etc.)

- Python is an interpreted language, First, Python VM will do compilation (Syntax checking) and once syntax is verified, then only it is going to start execution. Python VM is having the built-in capability of compilation. We do not need to compile explicitly like in Java.
