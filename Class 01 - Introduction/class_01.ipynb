## **<span style="color: green;">3. Python is Dynamically Typed</span>**
<hr/>

- Languages such as **C** and **Java** are generally statically typed, where variable types are explicitly declared (or strongly type-checked at compile time). In Python, explicit type declaration for variables is not required.

- Python is a **dynamically typed** language, which means that you do not need to declare the type of a variable when you create it. Python automatically determines a variable’s type based on the value assigned to it. For example, if you assign an integer value to a variable, Python will treat it as an integer. If you later assign a string value to the same variable, Python will treat it as a string.

- Simple program to display the type of a variable in Python using the built-in `type()` function.


````python
a = 10
print(f'The value of a is {a} and its type is {type(a)}')
````

Output:
```text
The value of a is 10 and its type is <class 'int'>
```

- Have a look at the following example to understand dynamic typing in Python. This program declares the variables `a` and `b`, assigns them integer values, and then reassigns a string value to variable `a` without any type declaration or errors.

````python
# Declaring the variables a, b and assigning initial values as 10 and 20 accordingly. Which are of type integer.
a, b = 10, 20
print(f'Initially, a = {a} and b = {b}') # Semicolon is not required at the end of statements in Python
print(f'Types of a and b are {type(a)} and {type(b)} respectively.')

# Reassigning a string value to variable a. This is possible in Python because it is dynamically typed.
a = "Now I am a string!"
print(f'After reassignment, a = {a} and b = {b}')
print(f'Types of a and b are {type(a)} and {type(b)} respectively.')
````

Output:
```text
Initially, a = 10 and b = 20
Types of a and b are <class 'int'> and <class 'int'> respectively.
After reassignment, a = Now I am a string! and b = 20
Types of a and b are <class 'str'> and <class 'int'> respectively.
```

## **<span style="color: green;">4. Why Python was named as Python</span>**
<hr/>

- Learning Python is fun. The language name “Python” was inspired by the BBC comedy show **Monty Python’s Flying Circus** (1969–1974), which Guido van Rossum liked.

- The name was chosen to be short, unique, and slightly mysterious, reflecting the language's emphasis on simplicity and creativity. It has no connection to the snake of the same name.

- Extra fact - **Hadoop**, The name "Hadoop" was inspired by a toy elephant owned by the son of one of the creators, Doug Cutting. It was chosen because it was unique and easy to remember, similar to how Python's name was chosen for its uniqueness and memorability.

## **<span style="color: green;">5. Python: Design Influences and Programming Paradigms</span>**
<hr/>

- Python was designed with ideas influenced by multiple languages and programming traditions:

    1. **ABC**
    A major influence on Python’s readability, simplicity, and clean syntax.

    2. **C (and, to a lesser extent, C++)**
    Influence on syntax style and implementation approach (CPython is implemented in C).

    3. **Modular language traditions (including Modula-3)**
    Influence on module-based program organization and software design.

    4. **Perl and shell scripting traditions**
    Influence on scripting convenience and practical text-processing workflows.

    5. **Functional programming traditions (e.g., Lisp)**
    Influence on first-class functions, `lambda`, `map`, and `filter`.

    6. **Object-oriented traditions (e.g., Smalltalk)**
    Conceptual influence on object-oriented design in Python.

- Therefore, Python is a multi-paradigm language that supports procedural, object-oriented, functional, scripting, and modular programming styles. It was designed to be a general-purpose language that can be used for a wide variety of applications, and its design reflects influences from many different programming traditions. Simply, Python consists of a blend of features and ideas from various languages, making it versatile and adaptable to different programming needs.
